```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
```

```python
df =fetch_california_housing()
dataset=pd.DataFrame(df.data)
dataset.columns=df.feature_names
dataset['Price']=df.target
dataset.head()
```

```python
print(dataset)
```

```python
#Independent feature and dependent features
x=dataset.drop('Price',axis=1) # Drop the target variable from features
y=df.target
```

```python
y
```

```python
## train and test split
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x, y, test_size=0.30,random_state=42)
```

```python
x_train
```

```python
## standarizing data set
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
x_train
```

## ***Import and Implimentation of linear regression model***

```python
from sklearn.linear_model import LinearRegression
## cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
regression=LinearRegression()
regression.fit(x_train,y_train)
mse=cross_val_score(regression,x,y,scoring='neg_mean_squared_error',cv=10)
mean_mse=np.mean(mse)
print(mean_mse)
```

```python
## prediction
reg_pred= regression.predict(x_test)
print(reg_pred)
```

```python
import seaborn as sns
sns.kdeplot(reg_pred - y_test)
plt.show()
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,reg_pred)
print(score)
```

# **Give New Input For Value prediction**

```python
# New house input
new_house = [[
    8.0,       # MedInc
    30.0,      # HouseAge
    6.0,       # AveRooms
    1.0,       # AveBedrms
    1500.0,    # Population
    3.0,       # AveOccup
    34.0,      # Latitude
    -118.0     # Longitude
]]

# Scale the new input using the already-fitted scaler
new_house_scaled = scaler.transform(new_house)

# Predict
predicted_price = regression.predict(new_house_scaled)

print("Predicted House Price:", predicted_price[0])
print("Predicted Price in Dollars: $", predicted_price[0] * 100000)
```
